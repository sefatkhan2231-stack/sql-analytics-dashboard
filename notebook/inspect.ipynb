```python
from pathlib import Path
import pandas as pd
```

```python
PROJECT_DIR = Path.cwd().parent
DATA_PATH = PROJECT_DIR / 'dataset' / 'Sample - Superstore.csv'
```

```python
df = pd.read_csv(DATA_PATH, encoding='latin-1')

df.head()
```

```python
print("\n--- Dataset Shape ---")
print(df.shape)

print("\n--- Duplicate Check ---")
print(df.duplicated().sum())

print("\n--- Missing Value ---")
print(df.isnull().sum())

print("\n--- Columns Name ---")
print(df.columns.to_list())

print("\n--- DATA TYPES AND NON-NULL COUNTS ---")
print(df.info())

print("\n--- Numeric Summery ---")
print(df.describe())
```

- There are 9994 rows and 21 columns present in this dataset.
- There are no duplicate and null values.
- There are 3 float, 3 integer and 15 object type of data.
- After describing it we did not find any visible outlier yet, but we are not that sure about it we need further observation.

```python
for col in ['Order Date', 'Ship Date']:
    df[col] = pd.to_datetime(df[col], errors='coerce')

print("\n--- Date Validation ---")
print(f'Invalid Order Date:', df['Order Date'].isna().sum())
print(f'Invalid Ship Date:', df['Ship Date'].isna().sum())

invalid_date = df['Ship Date'] < df['Order Date']
print(f'\n Invalid Date:', invalid_date.sum())

print("\n--- Numeric Validation ---")
print(f'Sales <= 0:', (df['Sales'] <= 0).sum())
print(f'Quantity <= 0:', (df['Quantity'] <= 0).sum())
print(f'Discount < 0:', (df['Discount'] < 0).sum())
print(f'Discount > 1:', (df['Discount'] > 1).sum())

print("\n--- Profit Validation ---")
print(f'Profit < 0:', (df['Profit'] < 0).sum())
print(f'Profit == 0:', (df['Profit'] == 0).sum())

print("\n--- ID Validation ---")

for col in ['Row ID', 'Order ID', 'Customer ID', 'Product ID']:
    print(f'\n{col}')
    print('Missing:', df[col].isna().sum())
    print('Unique values:', df[col].nunique())
    print('Duplicate values:', df[col].duplicated().sum())

print("\n--- Postal Code Validation ---")
print(f'Postal Code < 0:', (df['Postal Code'] < 0).sum())

print("\n--- Order Line Analysis ---")
lines_per_order = df.groupby('Order ID').size()

print('Orders with multiple lines:', 
      (lines_per_order > 1).sum())
print('Maximum rows in an order:', lines_per_order.max())

print('\n--- Five Lowest-Profit Rows ---')
print(df.nsmallest(5, 'Profit')[
    ['Order ID', 'Order Date', 'Customer ID', 'Product Name', 'Sales', 'Discount', 'Quantity', 'Profit']
].to_string(index=False))

print('\n--- Postal Code Check ---')
print('Postal code data type:', df['Postal Code'].dtype)
print('Unique postal codes:', df['Postal Code'].nunique())
print(df['Postal Code'].sort_values().unique()[:20])
print('Maximum:', df['Postal Code'].max())
print('Minimum:', df['Postal Code'].min())
print('Postal codes beginning with zero may need review.')
```
